```python
import pandas as pd
import sqlite3

# connect database
conn = sqlite3.connect(r"C:\Users\VIGNESH V\OneDrive\Desktop\python\testdb.db")
cur = conn.cursor()

# data
stock_data = {
    "bijili 50pcs": {"in_qty": 5, "out_qty": 2},
    "bijili 100pcs": {"in_qty": 200, "out_qty": 40},
    "2 inch 3pcs": {"in_qty": 300, "out_qty": 170},
    "fp spl": {"in_qty": 80, "out_qty": 30},
    "fp dlx": {"in_qty": 100, "out_qty": 100}
}

# create table
cur.execute("""
CREATE TABLE IF NOT EXISTS stock (
    item_name TEXT,
    in_qty INTEGER,
    out_qty INTEGER
)
""")

# insert data (safe way)
for key, value in stock_data.items():
    cur.execute(
        "INSERT INTO stock (item_name, in_qty, out_qty) VALUES (?, ?, ?)",
        (key, value["in_qty"], value["out_qty"])
    )

conn.commit()

# read into pandas
df = pd.read_sql("SELECT * FROM stock", conn)

conn.close()
df
```

```python
class bill:
    def __init__(self,name,qty,price):
        self.name=name
        self.qty=qty
        self.price=price
    def calcutale(self):
        return (self.qty*self.price)
    def dobill(self):
        itemname=self.name
        total_bill=self.calcutale()
        return total_bill
    
per1=bill("Apple",0.5,240)
per1.calcutale()
per1.dobill()
```

```python
class discount(bill):
    def total(self):
        total=super().dobill()
        discounted=total*0.95
        return discounted
```

```python
from fastapi import FastAPI, Request
from fastapi.responses import JSONResponse
import mysql.connector

app = FastAPI()

def get_db():
    return mysql.connector.connect(
        host="localhost",
        user="root",
        password="Vignesh@#3",
        database="testdb"
    )

# ---------------- GET USERS ----------------
@app.get("/getdata")
def get_users():
    conn = get_db()
    cursor = conn.cursor(dictionary=True)
    cursor.execute("SELECT * FROM items")
    users = cursor.fetchall()
    conn.close()
    return users


# ---------------- ADD USER ----------------
@app.post("/addusers")
async def add_user(request: Request):
    data = await request.json()
    name = data["name"]

    conn = get_db()
    cursor = conn.cursor()
    cursor.execute(
        "INSERT INTO users (name) VALUES (%s)",
        (name,)
    )
    conn.commit()
    conn.close()

    return {"message": "User has been added successfully!"}


# ---------------- EDIT USER ----------------
@app.post("/editusers")
async def edit_user(request: Request):
    data = await request.json()
    user_id = int(data["id"])
    name = data["name"]

    conn = get_db()
    cursor = conn.cursor()
    cursor.execute(
        "UPDATE users SET name=%s WHERE id=%s",
        (name, user_id)
    )
    conn.commit()
    conn.close()

    return {"message": "Updated successfully!"}


# ---------------- DELETE USER ----------------
@app.post("/deleteusers")
async def delete_user(request: Request):
    data = await request.json()
    user_id = int(data["id"])

    conn = get_db()
    cursor = conn.cursor()
    cursor.execute(
        "DELETE FROM users WHERE id=%s",
        (user_id,)
    )
    conn.commit()
    conn.close()

    return {"message": "User has been deleted successfully"}
```
